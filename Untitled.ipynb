```python
import os
import sys
import json
import time
from pathlib import Path

# Force-enable the opt-in neural ST-PIGNN SHAP engine built by Codex
os.environ["TOXICITY_INCLUDE_NEURAL_EXPLANATION"] = "1"

# Fix relative imports for Jupyter environment context
root_path = Path(os.getcwd()).resolve()
if str(root_path) not in sys.path:
    sys.path.insert(0, str(root_path))

import torch
from router.api.routes import route  
from shared.schemas import RouteRequest
from router.api.main import app

# Target Hardware Alignment (Tesla V100 Check)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("=" * 90)
print(f"HARDWARE INITIALIZATION: Utilizing {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}")
print("=" * 90)

# Safe attribute assignment checking via try/except block
if device.type == "cuda" and hasattr(app, "state"):
    try:
        # Check if model exists inside state without triggering Starlette's KeyError mapping
        if "model" in app.state._state and app.state._state["model"] is not None:
            app.state._state["model"] = app.state._state["model"].to(device)
            print("ST-PIGNN Tensor Layers successfully mapped to Tesla V100 HBM2 Memory Layer.")
        else:
            print("Model lifecycle managed dynamically by internal router dependencies. Proceeding to simulation...")
    except Exception as e:
        print(f"Note on explicit mapping layout: {e}. Handing tensor placement over to system defaults.")

# Define Simulation Variables
modes = ["jogger", "cyclist", "two_wheeler", "car"]
base_payload = {
    "origin": [12.97530, 77.60660],      # MG Road Snapping Point
    "destination": [13.02190, 77.56710], # IISc Side Snapping Point
}

def print_separator(title):
    print("\n" + "─" * 90)
    print(f"• {title}")
    print("─" * 90)

async def run_notebook_replication():
    print(f"Simulation Frame Load: Historical Trace Configuration")
    saved_responses = {}

    # --- STAGE 1: FORWARD INFERENCE & ROUTING ---
    print_separator("STAGE 1: Multi-Mode Spatiotemporal Simulations")
    for mode in modes:
        start_time = time.time()
        try:
            # Wrap payload parameters cleanly in the Pydantic RouteRequest object the endpoint expects
            req_object = RouteRequest(
                origin=base_payload["origin"],
                destination=base_payload["destination"],
                mode=mode
            )
            
            # Execute the endpoint directly through native async await
            res_data = await route(req_object)
            elapsed = time.time() - start_time
            print(f"├── {mode.upper():<12} Pass: status=200 | computed in {elapsed:.2f}s")
            
            # Extract response model to structured dict
            if hasattr(res_data, "model_dump"):
                saved_responses[mode] = res_data.model_dump()
            else:
                saved_responses[mode] = res_data.dict()
                
        except Exception as e:
            print(f"❌ {mode.upper():<12} Calculation Fault: {e}")

    # --- STAGE 2: EXPLANATIONS & SHAP ATTRIBUTIONS ---
    print_separator("STAGE 2: Integrated Neural Model SHAP Explanations")
    for mode in modes:
        data = saved_responses.get(mode, {})
        candidates = data.get("candidates", [])
        if not candidates:
            continue
        
        rec = next((c for c in candidates if c.get("recommended")), candidates[0])
        print(f"\n[Commute Mode: {mode.upper()}] Allocated Corridor ID: {rec['id']} (Rank {rec['preference_rank']})")
        print(f" ├── Route Score Attribution Payload:")
        print(json.dumps(rec["explanation"]["route_score"], indent=6))
        
        neural = rec["explanation"].get("neural_model")
        if neural and neural.get("available"):
            print(f" └── Neural SHAP Top Feature Attributions (ST-PIGNN Layer):")
            print(json.dumps(neural["feature_attributions"], indent=6))
        else:
            print(f" └── Neural SHAP Layer: Unavailable ({neural.get('reason') if neural else 'No Data'})")

    # --- STAGE 3: GALE-SHAPLEY MITIGATION AND CORRIDOR DEVIATIONS ---
    print_separator("STAGE 3: Gale-Shapley Spatial Node Deviations")
    for mode in modes:
        data = saved_responses.get(mode, {})
        candidates = data.get("candidates", [])
        if len(candidates) < 2:
            print(f"• {mode.upper()}: Insufficient corridor candidates to evaluate delta paths.")
            continue
            
        # Analyze topological deviation caused by game equilibrium matching
        baseline_nodes = set(candidates[0]["node_ids"])
        recommended_candidate = next((c for c in candidates if c.get("recommended")), candidates[0])
        recommended_nodes = set(recommended_candidate["node_ids"])
        
        swapped_out = baseline_nodes - recommended_nodes
        swapped_in = recommended_nodes - baseline_nodes
        
        print(f"\n🔹 Game Equilibrium Deflection Status [{mode.upper()}]:")
        if recommended_candidate["preference_rank"] > 1:
            print(f" ⚠️ SYSTEM NOTICE: Diverted from Rank 1 due to Gale-Shapley Capacity Congestion Control!")
        print(f" ├── High-Exposure Nodes Dropped : {len(swapped_out)}")
        print(f" ├── Clean Corridors Nodes Added : {len(swapped_in)}")
        if swapped_in:
            print(f" └── Mitigated Graph Intersections (Sample): {list(swapped_in)[:3]}")
        else:
            print(f" └── Path Optimization: Remaining on raw distance minimum pathway.")

# Execute the pipeline using ambient notebook loop
await run_notebook_replication()
```
